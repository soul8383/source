```python
import pandas as pd


df = pd.DataFrame({ "id": ["S01", "S02", "S03", "S04"],
"gpa": [3.8, 2.1, 3.2, 1.9],
"dropout": [0, 1, 0, 1],
"credits": [120, 60, 90, 45] })
records = df.to_dict("records")


records = df.to_dict("records")
print([r["id"] for r in sorted(records, key=lambda r: r["gpa"])])
labels = df["dropout"].map({ 0: "비탈락", 1: "탈락" })
print(labels.head(2).tolist())


class StudentAnalyzer:
  def __init__(self, df):
    self._df= df
  def dropout_rate(self):
    data = self._df["dropout"].dropna()
    data = pd.to_numeric(data, errors="coerce")
    data = data.dropna()
    return data.mean()

analyzer = StudentAnalyzer(df)
print(analyzer.dropout_rate())
```

```python
class LogisticModel:
  def predict(self, X):
    return "로지스틱회귀예측"

class RandomForestModel:
  def predict(self, X):
    return "랜덤포레스트예측"
class XGBoostModel:
  def predict(self, X):
    return "XGBoost예측"

models = [
LogisticModel(),
RandomForestModel(),
XGBoostModel()]

for model in models:
  print(model.predict("학생데이터"))
```
